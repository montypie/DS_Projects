```python
# Single linkage drawback demo

import matplotlib.pyplot as plt
import numpy as np
from scipy.cluster.hierarchy import linkage, dendrogram

# Functie om een reeks punten te genereren die het "chain effect" demonstreren
def generate_chain_effect_data():
    np.random.seed(0)

    # Drie compacte clusters
    cluster_A = np.random.normal(loc=[0, 0], scale=0.2, size=(20, 2))
    cluster_B = np.random.normal(loc=[5, -2], scale=0.2, size=(20, 2))
    cluster_C = np.random.normal(loc=[10, 0], scale=0.2, size=(20, 2))

    # Punten die een keten vormen tussen de clusters
    chain = np.column_stack((np.linspace(0, 10, 15), np.sin(np.linspace(0, 10, 15))))
    
    # Samenvoegen van alle punten
    points = np.vstack([cluster_A, cluster_B, cluster_C, chain])
    return points

# Punten genereren
points = generate_chain_effect_data()

# Single Linkage Clustering uitvoeren
Z = linkage(points, method='single')

# Plot de punten en de dendrogram
plt.figure(figsize=(15, 5))

# Plot de punten
plt.subplot(1, 2, 1)
plt.scatter(points[:, 0], points[:, 1], s=30)
plt.title("Gegenereerde Punten")

# Plot de dendrogram
plt.subplot(1, 2, 2)
dendrogram(Z)
plt.title("Dendrogram - Single Linkage Clustering")

plt.tight_layout()
plt.show()
```
