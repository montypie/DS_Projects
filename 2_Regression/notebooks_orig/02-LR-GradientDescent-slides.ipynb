```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
import os

from utils.visualization import BLUE, GREEN, standard_setup
standard_setup()
matplotlib.use('Agg') # Comment out this line if you want to display the plots on the screen
```

## Functions

```python
def generate_thetas(X, y, learning_rate = 0.1, n_iterations = 10, random_seed = 42):

    np.random.seed(random_seed)
    # Add a column of ones to X for the intercept term
    X_b = np.c_[np.ones((100, 1)), X]  # Add x0 = 1 to each instance

    # Gradient Descent
    m = len(X_b)
    theta = np.random.randn(2, 1)  # Random initialization

    thetas = [theta]

    for iteration in range(n_iterations):
        gradients = (2/m) * X_b.T.dot(X_b.dot(theta) - y)
        theta = theta - learning_rate * gradients
        # Print the intermediate results
        thetas.append(theta)
    return thetas

def compute_mse_loss(X_b, y, theta):
    """
    Compute the Mean Squared Error (MSE) loss for a linear regression model.

    Parameters:
    X_b (np.ndarray): The input features matrix with a column of ones (for bias).
    y (np.ndarray): The target values vector.
    theta (np.ndarray): The model parameters (weights), including the intercept.

    Returns:
    float: The MSE loss computed for the model parameters.
    """
    m = len(X_b)
    predictions = X_b.dot(theta)
    errors = predictions - y
    mse_loss = (1/m) * np.sum(errors ** 2)
    return mse_loss

```

## Locaties

```python
loc_figs = 'figures'
```

### Dataset maken
#### De X/y values

```python
# Generate synthetic dataset
np.random.seed(42)  # For reproducibility
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)
```

### De contour plot zijn values

```python
# Define the grid range for theta_0 and theta_1
theta_0_min, theta_0_max = -5, 12
theta_1_min, theta_1_max = -5, 12

# Create a grid of values
theta_0_vals = np.linspace(theta_0_min, theta_0_max, 400)
theta_1_vals = np.linspace(theta_1_min, theta_1_max, 400)
theta_0_grid, theta_1_grid = np.meshgrid(theta_0_vals, theta_1_vals)

# Compute the MSE loss for each combination of theta_0 and theta_1
X_b = np.c_[np.ones((100, 1)), X]
mse_losses = np.array([
    compute_mse_loss(X_b, y, np.array([[theta_0], [theta_1]]))
    for theta_0, theta_1 in zip(np.ravel(theta_0_grid), np.ravel(theta_1_grid))
]).reshape(theta_0_grid.shape)
```

### Een lineaire regressie die goed werkt

```python
errors = {'goed' : list(),
            'hoog' : list(),
            'laag' : list()}
```

```python
thetas = generate_thetas(X, y, learning_rate = 0.15, n_iterations = 15)


for n, theta in enumerate(thetas):
    fig, axs = plt.subplots(1, 2, figsize = (14, 6))
    axs[0].scatter(X, y, color = BLUE)
    xvals = [X.min(), X.max()]
    axs[0].plot(xvals, [theta[0] + theta[1]*x for x in xvals], color=GREEN)

    axs[0].set_xlabel("Feature waarde")
    axs[0].set_ylabel("Label waarde")
    axs[0].set_xlim([-0.1, 2.1])
    axs[0].set_ylim([0, 12])

    cp = axs[1].contour(theta_0_grid, theta_1_grid, mse_losses, levels=np.logspace(-2, 3, 20), cmap='viridis')
    for theta_contour in thetas[:n+1]:
        axs[1].scatter(theta_contour[0], theta_contour[1], color = GREEN, s = 35)
        errors['goed'].append(np.mean((theta[0] + theta[1] * X - y) ** 2))
    axs[1].set_xlabel(r"$\theta_0$")
    axs[1].set_ylabel(r"$\theta_1$")
    axs[1].set_xlim([-5, 12])
    axs[1].set_ylim([-5, 12])

    if n < 10:
        n_write = f'0{n}'
    else:
        n_write = str(n)
    fig.savefig(os.path.join(loc_figs, 'goede_lr', f"iter{n_write}.png"))
```
