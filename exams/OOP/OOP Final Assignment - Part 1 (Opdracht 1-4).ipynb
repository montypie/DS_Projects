## Deel 1: Basic Classes en Instance Methods (30 minuten)

**Theorie:** Methods, Instance methods, `__init__`, `self`

### Opdracht 1.1: Book Class

Maak een `Book` class met de volgende specificaties:

**Attributen:**
- `title` (str): De titel van het boek
- `author` (str): De auteur van het boek
- `isbn` (str): Het ISBN nummer (unieke identificatie)
- `publication_year` (int): Het publicatiejaar
- `available` (bool): Of het boek beschikbaar is (standaard True)

**Instance Methods:**
- `__init__()`: Initialiseer alle attributen
- `display_info()`: Print alle informatie van het boek in een leesbaar formaat
- `borrow()`: Zet `available` op False als het boek beschikbaar is, anders print een waarschuwing
- `return_book()`: Zet `available` terug op True

**Test je code:**
```python
book1 = Book("1984", "George Orwell", "978-0451524935", 1949)
book1.display_info()
book1.borrow()
book1.borrow()  # Zou een waarschuwing moeten geven
book1.return_book()
```

````python
# OPDRACHT 1.1: Schrijf hier je Book class
class Book:

    def __init__(self, title, author, isbn, publication_year):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.publication_year = publication_year
        self.available = True

    def display_info(self):
        info = f"Book: '{self.title}' by {self.author}, isbn-number: {self.isbn}, published in {self.publication_year}"
        print(info)

    def borrow(self):
        if self.available == True:
            self.available = False
        else:
            print("Warning: the book is unavailable")

    def return_book(self):
        self.available = True
````

````python
# Test code voor Opdracht 1.1
book1 = Book("1984", "George Orwell", "978-0451524935", 1949)
book1.display_info()
book1.borrow()
book1.borrow()  # Zou een waarschuwing moeten geven
book1.return_book()
````

### Opdracht 1.2: Member Class

Maak een `Member` class voor bibliotheekleden:

**Attributen:**
- `name` (str): Naam van het lid
- `member_id` (str): Uniek lidnummer
- `email` (str): Email adres
- `borrowed_books` (list): Lijst van geleende boeken (start leeg)

**Instance Methods:**
- `__init__()`: Initialiseer alle attributen
- `display_info()`: Print alle informatie van het lid
- `get_borrowed_count()`: Return het aantal geleende boeken

**Test je code:**

### Opdracht 2.1: Specialisatie van Books

Niet alle boeken zijn hetzelfde! Maak gespecialiseerde book classes:

**A. `EBook` class (erft van `Book`):**
- Extra attribuut: `file_size` (float, in MB)
- Extra attribuut: `format` (str, bijv. "PDF", "EPUB")
- Override `display_info()` om ook file size en format te tonen
- Extra method: `download()` - print een bericht dat het boek wordt gedownload

**B. `AudioBook` class (erft van `Book`):**
- Extra attribuut: `narrator` (str)
- Extra attribuut: `duration_minutes` (int)
- Override `display_info()` om ook narrator en duration te tonen
- Extra method: `play()` - print een bericht dat het audioboek wordt afgespeeld

**C. `PhysicalBook` class (erft van `Book`):**
- Extra attribuut: `shelf_location` (str, bijv. "A-5")
- Extra attribuut: `condition` (str, bijv. "New", "Good", "Fair")
- Override `display_info()` om ook location en condition te tonen

**Test je code:**
```python
ebook = EBook("Python Programming", "John Doe", "978-1234567890", 2023, 5.2, "PDF")
audiobook = AudioBook("The Great Gatsby", "F. Scott Fitzgerald", "978-0743273565", 1925, "Jake Gyllenhaal", 240)

```

````python
# OPDRACHT 2.1: Schrijf hier je EBook, AudioBook en PhysicalBook classes
# Kopieer eerst je Book class hierboven als deze nog nodig is
class EBook(Book):
    def __init__(self, title, author, isbn, publication_year, file_size, format):
        super().__init__(title, author, isbn, publication_year)
        self.file_size = file_size
        self.format = format

    def display_info(self):
        super().display_info()
        print(f"{' '*6}file_size: {self.file_size}, format: {self.format}")

    def download(self):
        print(f"'{self.title}' had been downloaded")
        
class AudioBook(Book):
    def __init__(self, title, author, isbn, publication_year, narrator, duration_minutes):
        super().__init__(title, author, isbn, publication_year)
        self.narrator = narrator
        self.duration_minutes = duration_minutes

    def display_info(self):
        super().display_info()
        print(f"{' '*6}narrator: {self.narrator}, duration: {self.duration_minutes}")

    def play(self):
        print(f"'{self.title}' is playing now")


class PhysicalBook(Book):
    def __init__(self, title, author, isbn, publication_year, shelf_location, condition):
        super().__init__(title, author, isbn, publication_year)
        self.shelf_location = shelf_location
        self.condition = condition

    def display_info(self):
        super().display_info()
        print(f"{' '*6}location: {self.shelf_location}, condition: {self.condition}")
````

````python
# Test code voor Opdracht 2.1
ebook = EBook("Python Programming", "John Doe", "978-1234567890", 2023, 5.2, "PDF")
audiobook = AudioBook("The Great Gatsby", "F. Scott Fitzgerald", "978-0743273565", 1925, "Jake Gyllenhaal", 240)
physical = PhysicalBook("To Kill a Mockingbird", "Harper Lee", "978-0061120084", 1960, "B-12", "Good")

ebook.display_info()
ebook.download()
print()
audiobook.display_info()
audiobook.play()
print()
physical.display_info()
````

### Opdracht 2.2: Specialisatie van Members

Verschillende soorten leden hebben verschillende rechten:

**A. `StudentMember` class (erft van `Member`):**
- Extra attribuut: `student_id` (str)
- Extra attribuut: `university` (str)
- Attribuut: `max_books` = 3 (studenten mogen max 3 boeken lenen)
- Override `display_info()` om ook student info te tonen

**B. `PremiumMember` class (erft van `Member`):**
- Extra attribuut: `membership_expiryiry` (str, datum)
- Attribuut: `max_books` = 10 (premium leden mogen max 10 boeken lenen)
- Extra method: `is_expired()` - simpele check (mag je zelf implementeren)
- Override `display_info()` om ook expiry date te tonen

**Test je code:**

````python
# OPDRACHT 2.2: Schrijf hier je StudentMember en PremiumMember classes
from datetime import datetime

class StudentMember(Member):
    def __init__(self, name, member_id, email, student_id, univ):
        super().__init__(name, member_id, email)
        self.student_id = student_id
        self.univ = univ
        self.max_books = 3

    def display_info(self):
        super().display_info()
        print(f"{' '*8}student_id: {self.student_id}, university: {self.univ}")

class PremiumMember(Member):
    def __init__(self, name, member_id, email, membership_expiry):
        super().__init__(name, member_id, email)
        self.membership_expiry = membership_expiry
        self.max_books = 10

    def is_expired(self):
        today = datetime.today().strftime('%Y-%m-%d')
        if self.membership_expiry <= today:
            return True
        else:
            return False

    def display_info(self):
        super().display_info()
        print(f"{' '*8}premium membership until: {self.membership_expiry}")
````

## Deel 3: Getters en Setters (45 minuten)

**Theorie:** Property decorators, `@property`, `@attribute.setter`, Encapsulation, Private attributes

### Opdracht 3.1: Bescherm Book Attributen

Pas je `Book` class aan met private attributen en properties:

**Vereisten:**
- Maak `isbn` een private attribuut (`__isbn`)
- Maak `publication_year` een private attribuut (`__publication_year`)
- Creëer een `@property` getter voor `isbn` (alleen lezen)
- Creëer een `@property` getter en setter voor `publication_year`
  - Setter validatie: jaar moet tussen 1000 en huidig jaar liggen
- Maak `available` een private attribuut met property en setter
  - Setter validatie: alleen boolean waarden toegestaan

**Test je code:**
```python
book = Book("Test Book", "Author", "978-1234567890", 2020)
print(book.isbn)  # Zou moeten werken
# book.isbn = "new-isbn"  # Zou fout moeten geven
book.publication_year = 2023  # Zou moeten werken
# book.publication_year = 3000  # Zou fout of waarschuwing moeten geven
```

````python
# OPDRACHT 3.1: Pas je Book class aan met getters en setters
# Je mag je eerdere Book class kopiëren en aanpassen

class Book:

    def __init__(self, title, author, isbn, publication_year):
        self.title = title
        self.author = author
        self.__isbn = isbn
        self.publication_year = publication_year
        self.available = True

    @property
    def isbn(self):
        return self.__isbn
    
    @property
    def publication_year(self):
        return self.__publication_year
    
    @property
    def available(self):
        return self.__available
    
    @available.setter
    def available(self, status):
        if isinstance(status, bool):
            self.__available = status
        else:
            print(f"Error: availability status must be True or False")
    
    @publication_year.setter
    def publication_year(self, year):
        today = datetime.today()
        if year >= 1000 and year <= today.year:
            self.__publication_year = year
        else:
            print(f"Error: year must be between 1000 and {today.year}")

    def display_info(self):
        info = f"Book: '{self.title}' by {self.author}, isbn-number: {self.__isbn}, published in {self.__publication_year}"
        print(info)

    def borrow(self):
        if self.__available == True:
            self.__available = False
        else:
            print("Warning: the book is unavailable")

    def return_book(self):
        self.__available = True
````

````python
# Test code voor Opdracht 3.1
book = Book("Test Book", "Author", "978-1234567890", 2020)
print(f"ISBN: {book.isbn}")  # Zou moeten werken
print(f"Publication year: {book.publication_year}")

# Test setter
book.publication_year = 2023
print(f"Updated year: {book.publication_year}")

# Uncomment om error te testen:
#book.isbn = "new-isbn"  # Zou fout moeten geven
book.publication_year = 3000  # Zou fout of waarschuwing moeten geven
book.publication_year = 2023
book.display_info()

# check availability setter:
print(book.available)
book.available = False
print(book.available)
book.available = 1
````

### Opdracht 3.2: Bescherm Member Attributen

Pas je `Member` class (en subclasses) aan:

**Vereisten:**
- Maak `email` een private attribuut met property en setter
  - Setter validatie: moet "@" bevatten (simpele email validatie)
- Maak `borrowed_books` een private attribuut
  - Creëer een `@property` getter die een **kopie** van de lijst teruggeeft (niet de originele lijst!)
  - Dit voorkomt dat externe code direct de lijst kan wijzigen
- Voor `StudentMember` en `PremiumMember`: maak `max_books` een class attribuut of property

**Test je code:**
```python
member = Member("Test User", "M999", "test@email.com")
print(member.email)
member.email = "newemail@test.com"  # Zou moeten werken
# member.email = "invalid-email"  # Zou fout of waarschuwing moeten geven
```

````python
# OPDRACHT 3.2: Pas je Member class aan met getters en setters
from copy import deepcopy

class Member:
    def __init__(self, name, memeber_id, email):
        self.name = name
        self.member_id = memeber_id
        self.__email = email
        self.__borrowed_books = []

    @property
    def email(self):
        return self.__email
    
    @email.setter
    def email(self, new_email):
        if "@" in new_email:
            self.__email = new_email
        else:
            print(f"Error: email must contain @")

    @property
    def borrowed_books(self):
        return deepcopy(self.__borrowed_books)

    def display_info(self):
        info = f"Member: {self.name}, id: {self.member_id}, email: {self.__email}"
        print(info)

    def get_borrowed_count(self):
        return len(self.borrowed_books)
    

class StudentMember(Member):

    def __init__(self, name, member_id, email, student_id, univ):
        super().__init__(name, member_id, email)
        self.student_id = student_id
        self.univ = univ

    @property
    def max_books(self):
        return 3

    def display_info(self):
        super().display_info()
        print(f"{' '*8}student_id: {self.student_id}, university: {self.univ}")

class PremiumMember(Member):
    def __init__(self, name, member_id, email, membership_expiry):
        super().__init__(name, member_id, email)
        self.membership_expiry = membership_expiry

    @property
    def max_books(self):
        return 10

    def is_expired(self):
        today = datetime.today().strftime('%Y-%m-%d')
        if self.membership_expiry <= today:
            return True
        else:
            return False

    def display_info(self):
        super().display_info()
        print(f"{' '*8}premium membership until: {self.membership_expiry}")
````

````python
# Test code voor Opdracht 3.2
member = Member("Test User", "M999", "test@email.com")
print(f"Email: {member.email}")
member.email = "newemail@test.com"  # Zou moeten werken
print(f"Updated email: {member.email}")

# Uncomment om error te testen:
member.email = "invalid-email"  # Zou fout of waarschuwing moeten geven

# Check (the amount of) borrowed books:
print(f"{member.name} has borrowed {member.get_borrowed_count()} books")
````

## Deel 4: Polymorfisme (45 minuten)

**Theorie:** Polymorfisme, Method overriding, Duck typing

### Opdracht 4.1: Universele Book Operations

Implementeer polymorf gedrag voor alle book types:

**Vereisten:**
- Zorg dat `EBook`, `AudioBook` en `PhysicalBook` allemaal een `borrow()` method hebben
  - `EBook.borrow()`: Verhoog een download counter (voeg attribuut toe)
  - `AudioBook.borrow()`: Verhoog een play counter
  - `PhysicalBook.borrow()`: Normale borrow functionaliteit (zoals in basis Book class)
- Alle book types moeten een `get_type()` method hebben die het type boek teruggeeft ("E-Book", "AudioBook", "Physical Book")

**Test je code met een lijst van verschillende book types:**
```python
books = [
    EBook("Python Guide", "Author A", "ISBN1", 2023, 5.0, "PDF"),
    AudioBook("History of Time", "Author B", "ISBN2", 2020, "Narrator X", 300),
    PhysicalBook("Novel", "Author C", "ISBN3", 2019, "C-7", "New")
]

for book in books:
    print(f"{book.title} - Type: {book.get_type()}")
    book.borrow()
```

````python
# OPDRACHT 4.1: Pas je book classes aan voor polymorfisme
class EBook(Book):
    def __init__(self, title, author, isbn, publication_year, file_size, format):
        super().__init__(title, author, isbn, publication_year)
        self.file_size = file_size
        self.format = format
        self.download_counter = 0

    def display_info(self):
        super().display_info()
        print(f"{' '*6}file_size: {self.file_size}, format: {self.format}")

    def download(self):
        print(f"'{self.title}' had been downloaded")

    def borrow(self):
        super().borrow()
        self.download_counter += 1

    def get_type(self):
        return "EBook"
        
class AudioBook(Book):
    def __init__(self, title, author, isbn, publication_year, narrator, duration_minutes):
        super().__init__(title, author, isbn, publication_year)
        self.narrator = narrator
        self.duration_minutes = duration_minutes
        self.play_counter = 0

    def display_info(self):
        super().display_info()
        print(f"{' '*6}narrator: {self.narrator}, duration: {self.duration_minutes}")

    def play(self):
        print(f"'{self.title}' is playing now")

    def borrow(self):
        super().borrow()
        self.play_counter += 1
    
    def get_type(self):
        return "AudioBook"
    
    
class PhysicalBook(Book):
    def __init__(self, title, author, isbn, publication_year, shelf_location, condition):
        super().__init__(title, author, isbn, publication_year)
        self.shelf_location = shelf_location
        self.condition = condition

    def display_info(self):
        super().display_info()
        print(f"{' '*6}location: {self.shelf_location}, condition: {self.condition}")

    def get_type(self):
        return "PhysicalBook"
````

````python
# Test code voor Opdracht 4.1
books = [
    EBook("Python Guide", "Author A", "ISBN1", 2023, 5.0, "PDF"),
    AudioBook("History of Time", "Author B", "ISBN2", 2020, "Narrator X", 300),
    PhysicalBook("Novel", "Author C", "ISBN3", 2019, "C-7", "New")
]

for book in books:
    print(f"{book.title} - Type: {book.get_type()}")
    book.borrow()
````

### Opdracht 4.2: Polymorf Member Gedrag

Implementeer een `can_borrow()` method voor alle member types:

**Vereisten:**
- Basis `Member`: kan lenen als minder dan 5 boeken zijn geleend
- `StudentMember`: kan lenen als minder dan 3 boeken zijn geleend
- `PremiumMember`: kan lenen als minder dan 10 boeken zijn geleend EN membership niet expired is

**Test je code:**
```python
members = [
    Member("User1", "M1", "user1@email.com"),
    StudentMember("User2", "M2", "user2@email.com", "S001", "University"),
    PremiumMember("User3", "M3", "user3@email.com", "2025-12-31")
]

for member in members:
    print(f"{member.name} can borrow: {member.can_borrow()}")
```

````python
# OPDRACHT 4.2: Implementeer can_borrow() voor alle member types

class Member:
    def __init__(self, name, memeber_id, email):
        self.name = name
        self.member_id = memeber_id
        self.__email = email
        self.__borrowed_books = []

    @property
    def max_books(self):
        return 5

    @property
    def email(self):
        return self.__email
    
    @email.setter
    def email(self, new_email):
        if "@" in new_email:
            self.__email = new_email
        else:
            print(f"Error: email must contain @")

    @property
    def borrowed_books(self):
        return deepcopy(self.__borrowed_books)

    def display_info(self):
        info = f"Member: {self.name}, id: {self.member_id}, email: {self.__email}"
        print(info)

    def get_borrowed_count(self):
        return len(self.borrowed_books)
    
    def can_borrow(self):
        if len(self.borrowed_books) < self.max_books:
            return True
        else:
            return False
    

class StudentMember(Member):

    def __init__(self, name, member_id, email, student_id, univ):
        super().__init__(name, member_id, email)
        self.student_id = student_id
        self.univ = univ

    @property
    def max_books(self):
        return 3

    def display_info(self):
        super().display_info()
        print(f"{' '*8}student_id: {self.student_id}, university: {self.univ}")


class PremiumMember(Member):
    def __init__(self, name, member_id, email, membership_expiry):
        super().__init__(name, member_id, email)
        self.membership_expiry = membership_expiry

    @property
    def max_books(self):
        return 10

    def is_expired(self):
        today = datetime.today().strftime('%Y-%m-%d')
        if self.membership_expiry <= today:
            return True
        else:
            return False

    def display_info(self):
        super().display_info()
        print(f"{' '*8}premium membership until: {self.membership_expiry}")

    def can_borrow(self):
        if self.get_borrowed_count() < self.max_books and not self.is_expired():
            return True
        else:
            return False
````

````python
# Test code voor Opdracht 4.2 (changed the expiration date for User3 to test borrow_books prop)
members = [
    Member("User1", "M1", "user1@email.com"),
    StudentMember("User2", "M2", "user2@email.com", "S001", "University"),
    PremiumMember("User3", "M3", "user3@email.com", "2025-12-31")
]

for member in members:
    print(f"{member.name} can borrow: {member.can_borrow()}")

members[2].membership_expiry = "2025-10-31"

print("----------Check after updates------------")
for member in members:
    print(f"{member.name} can borrow: {member.can_borrow()}")
````
