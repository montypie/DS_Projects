## Deel A: GUI Interface - Gebruikshandleiding

### Introductie

Voor Deel 3 heb je een **grafische gebruikersinterface (GUI)** gekregen om je Library Management System te testen. Deze GUI gebruikt alle `.py` files die je hebt gemaakt en maakt het mogelijk om alle functionaliteit interactief te testen.

---

### De GUI Starten

Om de GUI te starten, voer het volgende commando uit in de terminal (zorg dat je in de juiste map bent):

```powershell
python library_gui.py
```

De GUI opent automatisch met voorbeelddata (boeken en leden) zodat je meteen kunt beginnen met testen.

---

### Interface Overzicht

De GUI bestaat uit **5 hoofdtabs**, elk met specifieke functionaliteit:

---

#### 1. 📚 Books Tab - Boeken Beheren

**Wat kun je hier doen:**
- **Boeken bekijken:** Links zie je een lijst van alle boeken in de bibliotheek
- **Nieuw boek toevoegen:** Rechts kun je een nieuw boek toevoegen
  - Kies het type: Physical, EBook, of AudioBook
  - Vul de basisgegevens in (titel, auteur, ISBN, jaar)
  - Vul de type-specifieke velden in (deze veranderen automatisch)
  - Klik op "Add Book"

**Type-specifieke velden:**
- **Physical Book:** Shelf Location, Condition
- **EBook:** File Size (MB), Format (PDF/EPUB/etc.)
- **AudioBook:** Narrator, Duration (minutes)

**Extra functie:**
- Selecteer een boek en klik op "View Selected Book Details" voor uitgebreide informatie

#### 2. 👥 Members Tab - Leden Beheren

**Wat kun je hier doen:**
- **Leden bekijken:** Links zie je alle geregistreerde leden
- **Nieuw lid toevoegen:** Rechts kun je een nieuw lid registreren
  - Kies het type: Regular, Student, of Premium
  - Vul naam, member ID, en email in
  - Gebruik "Generate ID" om automatisch een uniek ID te genereren
  - Vul type-specifieke velden in voor Student of Premium leden

**Type-specifieke velden:**
- **Student:** Student ID, University
- **Premium:** Membership Expiry Date (format: YYYY-MM-DD)

**Let op:** Elk type lid heeft een ander maximum aantal boeken:
- Regular: 5 boeken
- Student: 3 boeken
- Premium: 10 boeken

#### 3. 📝 Transactions Tab - Lenen & Terugbrengen

**Wat kun je hier doen:**

**Boek lenen:**
1. Vul Member ID in (bijv. M001)
2. Vul Book ISBN in
3. Controleer/pas de datum aan
4. Klik op "Borrow Book"

**Boek terugbrengen:**
1. Vul Member ID in
2. Vul Book ISBN in
3. Vul de return date in
4. Klik op "Return Book"

**Transactiehistorie:**
- Onderaan zie je alle transacties (actief en voltooid)
- ✓ Returned = boek is teruggebracht
- Active = boek is nog uitgeleend

**Belangrijke validaties die worden getest:**
- Kan het lid nog meer boeken lenen? (max limiet)
- Is het boek beschikbaar?
- Bestaat het lid/boek in het systeem?
- Is het Premium membership verlopen?

#### 4. 🔍 Search Tab - Zoekfunctionaliteit

**Wat kun je hier doen:**
- Voer een zoekterm in (naam, titel, ISBN, email, etc.)
- Klik op "Search Books & Members"
- De interface zoekt **automatisch** in:
  - Alle boektitels, auteurs, en ISBNs
  - Alle ledennamen, IDs, en emails

**Dit test:**
- De `Searchable` interface
- De `matches_search()` methode
- De `get_search_keywords()` methode

#### 5. ⚙️ Advanced Features Tab - Geavanceerde Testen

Deze tab test alle geavanceerde OOP concepten:

**Iterators Testen:**
- **"Iterate Through All Books"** - Test de `__iter__` en `__next__` methoden
- **"Show Available Books"** - Test de `available_books()` generator

**Generators Testen:**
- **"Find Books by Author"** - Test de `books_by_author()` generator
- **"Find Books by Type"** - Test de `books_by_type()` generator
- **"Show Active Members"** - Test de `active_members()` generator

**Utility Functions Testen:**
- **"Validate Email"** - Test `utils.validate_email()`
- **"Validate ISBN"** - Test `utils.validate_isbn()`
- **"Calculate Date Difference"** - Test `utils.calculate_days_between()`

Alle resultaten worden getoond in het resultatenveld onderaan.

## Deel B: Reflectievragen over OOP Concepten

Beantwoord de volgende vragen in eigen woorden. Deze vragen helpen je om dieper na te denken over de OOP concepten die je hebt toegepast in dit project.

### Vragen over Abstract Base Classes & Interfaces (Deel 5)

**Vraag 1:** Waarom kan je geen instantie maken van de `LibraryItem` class? Leg uit wat het verschil is tussen een abstracte class en een normale class.

**Antwoord Vraag 1:**

LibraryItem is een abstracte class. Die worden niet geinstantieert, maar dienen om een structuur (methodes en properties) van de afgeleide classes vast te leggen. De abstracte methodes en properties zijn dan verplicht in alle classes die van de abstract class erfen. Indien ze niet geimplementeerd worden, krijgt men een fout.

**Vraag 2:** De `Searchable` interface definieert twee abstracte methoden: `matches_search()` en `get_search_keywords()`. Waarom is het nuttig om deze als interface te definiëren in plaats van gewoon in elke class apart te implementeren?

**Antwoord Vraag 2:**

Het is nuttig om zeker te zijn dat alle classes die van deze interface erfen de implementatie van deze methodes voorzien, terwijl de concrete implementaties daarvan van class tot class kunnen vershillen.

Nochtans, alhoewel dit het geval is voor get_search_keywords() (in Book en Member zoekt men in verschillende attributes), is het me niet helemaal duidelijk waarom is matches_search() geen concrete ipv abstracte methode, want de code is identiek in beide classes en is niet afhankelijk van geen enkele attribute van een concrete class. Om de code niet te dupliceren, lijkt me best van Searchable een abstracte classe maken met de concrete methode matches_search().

**Vraag 3:** In het project implementeert `Book` zowel `LibraryItem` (abstract class) als `Searchable` (interface). Wat is het verschil tussen een abstract base class en een interface? 

**Antwoord Vraag 3:**

In theorie, een abstract class kan zowel abstracte als concrete methodes bevatten, terwijl een interface is een puur "contract" zonder concrete implementaties (alleen "wat" zonder "hoe"). In het praktijk is er in Python geen formeel vershil, eerder een conventie dat je in de docstring het doel van je classe duidelijk maakt: is het een interface of een abstracte base class is. Als men een concrete methode aan een interface toevoegt wordt het zogezegd een abstracte class. Dikwijls wordt er ook een I-prefix in de namen van interfaces gebruikt, zoals bv. "ISearchable".

### Vragen over Composition (Deel 6)

**Vraag 4:** In dit project gebruikt de `Library` class composition: het bevat `Book` en `Member` objecten in lijsten, en creëert `Transaction` objecten. Leg uit waarom dit composition is en geen inheritance. Wat zou het probleem zijn als `Library` zou erven van `Book`?

**Antwoord Vraag 4:**

Library is een composition omdat die met meerdere classes opereert, nm. Book, Member en Transaction, maar de methodes en properties van Library class zijn anders dan in die andere classes. Aangezien Library een verzameling van Books en Members is, zou het helemaal niet logisch zijn om van deze classes over te erfen: ze hebben weinig tot geen overeenkomsten in properties en gedrag. Terwijl als we composition gebruiken, kunnen we Library gemakkelijk met een andere type item uitbreiden, bv. DVD (voorbeeld uit verleden, weet ik).

**Vraag 5:** De `Transaction` class verbindt een `Member` en een `Book` object. Dit is een voorbeeld van composition waarbij één object verwijst naar andere objecten. Beschrijf een situatie in het echte leven waar je een soortgelijk patroon zou tegenkomen (een object dat andere objecten "bezit" of "gebruikt").

**Antwoord Vraag 5:**

Een delivery gebruikt vervoer en goederen; een bankoverschrijving gebruikt geld en rekeningen; een opleiding bezit een docent en studenten.

**Vraag 6:** In de `borrow_book()` methode wordt een nieuwe `Transaction` aangemaakt, en worden zowel het boek als het lid aangepast. Leg uit welke voordelen deze compositie-aanpak heeft voor het bijhouden van de geschiedenis. Wat zou moeilijker zijn zonder de `Transaction` class?

**Antwoord Vraag 6:**

Deze aanpaak zorgt ervoor dat er een permanente link ontstaat tussen de boeken en de mensen die ze uitlenen. Voor elk transactie wordt er een unique id aangewezen die dan kan gevonden worden wanneer een boek wordt teruggebracht. Zonder Transaction class zou onmogelijk zijn om te weten wie heeft wat uitgeleend, de return_book() zinvol te implementeren en de geschiedenis van de transacties bij te houden.

### Vragen over Iterators & Generators (Deel 7)

**Vraag 7:** De `Library` class implementeert `__iter__()` en `__next__()` om itereerbaar te maken. Leg uit in eigen woorden hoe dit patroon werkt. Wat gebeurt er wanneer je `for book in library` schrijft?

**Antwoord Vraag 7:**

___iter__() geeft het huidig object terug, __next__() bepaalt hoe aan het volgende object te geraken. Bv. om over alle boeken van de bib te itereren loop je een per een alle elementen uit de lijst van boeken die de class Library als property heeft.

**Vraag 8:** De methode `available_books()` gebruikt het `yield` keyword in plaats van `return`. Wat is het verschil tussen een generator (met yield) en een normale functie (met return)? Wanneer zou je een generator verkiezen boven een gewone functie die een lijst teruggeeft?

**Antwoord Vraag 8:**

yield geeft het huidige element terug, maar in tegenstelling tot return, wordt de functie niet gestopt, gewoon even onderbroken. Die herpakt zich wanneer het volgende element wordt gevraagd. Daarom moet je er zelf voor zorgen dat die stopt na dat alle elementen uitgeput zijn. Op die manier zit er telkens slechts 1 element in het geheugen en niet de volledige lijst (meestal efficienter).

**Vraag 9:** Vergelijk de `available_books()` generator met de `__iter__()` methode. Beide maken iteratie mogelijk, maar op verschillende manieren. Welke voordelen heeft elke aanpak? Wanneer zou je de ene kiezen boven de andere?

**Antwoord Vraag 9:**

`__iter__()` methode maakt de hele class itereerbaar, terwijl available_books() een gefilterde generator is die alleen beschikbare boeken yieldt.

In vergelijking met `__iter__()`, die over alle items itereert, zijn generators flexibeler — je kunt meerdere generators hebben met verschillende filters (beschikbare boeken, boeken per auteur, etc.) zonder de class zelf aan te passen. Generators zijn ook geheugen-efficiënter.

### Vragen over Modules & Scope (Deel 8)

**Vraag 10:** Het project is opgesplitst in meerdere modules (`library_items.py`, `members.py`, `library.py`, `config.py`, `utils.py`). Leg uit waarom dit een betere aanpak is dan alles in één groot bestand te zetten. 

**Antwoord Vraag 10:**

Een modulere aanpak is een van de fundamentele principes van OOP. Op die manier wordt de code in logische blokjes verdeeld en gelijkaardige functionaliteit bij elkaar gezet. Daarna wordt het geheel met die blokjes opgebouwd. Dit maakt het project flexiebeler en minder fout-gevoelig. Het is gemakkelijker blokjes toe te voegen/aanpassen/vervangen zonder de rest van de code aan te raken, lang in één groot bestand te zoeken of riskeren het geheel ineens kapot te maken. Ook kwa testing is het zinvoller.

**Vraag 11:** De `config.py` module bevat constanten zoals `MAX_BORROW_DAYS` en `DEFAULT_MAX_BOOKS`. Waarom is het een goede gewoonte om configuratiewaarden in een apart bestand te zetten? Wat zou er gebeuren als deze hardcoded waren in verschillende files?

**Antwoord Vraag 11:**

De algemene configuratieconstanten op één plaats te zetten is het gezonde verstand. Als je die op één plaats definieert moet je die ook slechts op één plaats aanpassen, indien nodig.

**Vraag 12:** In `utils.py` staat een variabele `_id_counters` met een underscore vooraan. Wat betekent deze conventie in Python? Waarom is het belangrijk om te begrijpen welke variabelen "privé" zijn en welke "publiek"?

**Antwoord Vraag 12:**

Ik heb geen variable "_id_counters" in utils.py, maar het onderschijd van privé en publieke variabelen is belangrijk voor een toegangscontrole. De al dan niet onderlijnde 'toegangsmodifiers' in Python bepalen welke attributes en methodes van een classe van buiten de classe, binnen de classe of door subclassen aangesproken kunnen worden. Ze helpen om properties en methoden veilig en georganiseerd te houden.

PS. Idealieter zou de functie generate_id() rekening moeten houden met de bestaande ids. Maar uit de Q&A sessie vanmorgen heb ik begrepen dat dit buiten de scope van dit oefening ligt.

### Algemene OOP Vragen (Reflectie)

**Vraag 13:** In dit project heb je polymorfisme toegepast: `Book`, `EBook`, `AudioBook`, en `PhysicalBook` hebben allemaal een `borrow()` methode, maar met verschillend gedrag. Leg uit hoe polymorfisme het mogelijk maakt om deze verschillende boektypes op een uniforme manier te behandelen in de `Library` class.

**Antwoord Vraag 13:**

De base class LibraryItem bepaalt dat er een borrow()-method in al zijn subclassen beschikbaar moet zijn. De concrete implementatie verschilt echter van item tot item. Polimorfisme principe laat toe om de logica en acties binnen de methode telkens afhankelijk van de concrete class te bepalen en tegelijk de gemenschappelijke naam behouden. Later worden die op een gelijkaardige manier opgeroepen in Library class (book.borrow()), maar de achterliggende acties zijn bij PhysicalBook, AudioBook en EBook zijn verschillend.
