 ### Opdracht 1: MONTY HALL

Het programma hieronder illustreert het phenomeen van voorwaardelijke kans bij het Monty Hall TV-spel.

```python
import random

def mh_simulation(doors, switch, n):
    """ Monty Hall TV-spel simulatie """

    bingos = 0

    for i in range(n):
        prize = random.choice(doors)
        door1 = random.choice(doors)
        open_d = [d for d in doors if d not in (door1, prize)]

        if switch == False: # option 1: pick the same door
            door2 = door1
        
        else: # option 2: switch the door
            other_d = [d for d in doors if d not in (door1, open_d[0])]
            door2 = other_d[0]
            
        if door2 == prize:
            bingos +=1
                
    return bingos / n * 100
```

Hulpfunctie om het aantal van simulaties (al dan niet door de gebruiker) in te stellen. 

```python
def nr_simulations():
    """ Set up number of times to run the simulation """

    n = input("Geef een geheel getal tussen 1 en 10000:")

    if n.isdigit() and int(n) > 0 and int(n) <= 10000:
        n = int(n)
    else:
        print("Ongeldige input! Het aantal van iteraties wordt automatisch gezet.")
        n = random.randint(1,10001)
        
    return n
```

START:
In het begin van het programma word er een input van de gebruiker gevraagd. Indien de input ongeldig is, wordt het aantal van simulaties (n) automatisch gezet. (Het aantal is hier gelimiteerd door 10000 om het programma niet te lang laten lopen.)

Er worden twee simulaties achter elkaar telkens n keer uitgevoerd -- één keer met de variable switch=False om bij dezelfde deur te blijven en dan met switch=True om van de deur te wisselen -- en het resultaat weergegeven.

```python
# Main program:
print("Welkom op de Monty Hall simulatie!")
print("Hoeveel keer wil je die uitvoeren?")

doors = list(range(1,4))
n = nr_simulations()
print(f"Elk scenario is {n} keer uitgevoerd met het volgende kans om prijs te winnen:")

# Run both scenarios and output the result
for switch in (False, True):
    odds = mh_simulation(doors, switch, n)
    if not switch:
        output = "* bij dezelfde deur: "
    else:
        output = "* bij het wisselen van deur: "
    output += f"{odds:.2f}%."
    print(output)
```

### Een mogelijke verklaring voor het resultaat:
Wanneer men de tweede keer bij zijn eerste keuze blijft is de kans op prijs dezelfde als wanneer men zou de "juiste" deur vanaf de eerste keer gokken, namelijk 1 op 3. Terwijl als men de tweede keer naar een andere deur omschakelt, kiest hij/zij uit minder deuren dan de eerste keer waardoor de kans om te winnen noteerbaar groter wordt. Hoe meer iteraties, hoe duidelijker het effect.
