### Opdracht 2: MASTER MIND

__Run All in this notebook to guess a 4-color code in 12 steps.__

Optionally you can set the number of guessing attepmts and colors in the code underneath.

Flow:
1. set all colors (optional per game: length of code, nr of attempts)
2. set random n from all colors > put in a sec_code list
3. initiate a counter = nr of attempts
4. get input (+ validate it!) > when ok:
5. compare in_colors 1 by 1 with the sec_colors
    - both color and index right > black peg
    - right color wrong index > white peg*
6. give feedback:
    - if all are black -> won! :)
    - if not and attempts counter == 0 -> lost :(
    - else: left nr of attempts (decrement counter by 1)
7. repeat from p.3 while counter > 0

--------------------
(*) take into account that colors can repeat!

```python
import random

def set_palettes():
    """ The one where the colors are set """

    # fancy circles
    all_colors = {"R": "\U0001F534",
                  "B": "\U0001F535",
                  "Y": "\U0001F7E1",
                  "G": "\U0001F7E2",
                  "O": "\U0001F7E0",
                  "P": "\U0001F7E3"
                  }

    # feedback tokens
    fb_colors = {"black": "\U00002B24", "white": "\U0001F785"}

    return (all_colors, fb_colors)

def set_code(colors, length):
    """ The one where computer chooses random color code """

    secret_code = []
    for i in range(length):
        secret_code.append(random.choice(colors))
        i += 1
    return secret_code

def make_guess(code_length):
    """ The one where the player makes a guess """

    code_in = input(f"Enter {code_length} color letters:")
    if not validate_input(code_in, code_length):
        print("Invalid input! Try again.")
        code_in = make_guess(code_length)
    upcode = code_in.upper()
    return upcode

def validate_input(code, length):
    """ The one where the input gets validated """

    if len(code) != length:
        return False
    allowed = "BGOPRYbgopry"
    if not set(code).issubset(set(allowed)):
        return False
    return True

def reveal_colors(color_code, palette):
    """ The one that prints colored circles """

    for c in color_code:
            print(palette[c], end="")

def get_feedback(code_in, sec_code):
    """ The one that compares input to the secret code """

    feedback = {"black": 0,
                "white": 0}
    amounts = {}
    for col in sec_code:
        amounts[col] = sec_code.count(col)

    for pos, color in enumerate(code_in):
            if color == sec_code[pos]:
                feedback["black"] += 1
                amounts[color] -= 1

    for pos, color in enumerate(code_in):
        if color in sec_code and amounts[color] > 0 and color != sec_code[pos]:
            feedback["white"] += 1
            amounts[color] -= 1
    
    return feedback
```

```python
# Main program

def play_mastermind(code_length, max_attempts):
    """ Master mind game """

    all_colors, fb_colors = set_palettes()
    sec_code = set_code([k for k in all_colors.keys()], code_length)
    steps = max_attempts

    print("WELCOME TO THE MASTER MIND GAME!")
    print(f"Try to crack a code of {code_length} colors in {max_attempts} steps.")
    print(f"The codes are:", end=" ")
    for c in all_colors:
        print(f"{c}{all_colors[c]}", end=" | ")
    print()

    while steps > 0:
        player_code = list(make_guess(code_length))
        print("\nYour guess: ", end="")
        reveal_colors(player_code, all_colors)
        feedback = get_feedback(player_code, sec_code)

        # game won: congratulate and stop
        if feedback["black"] == code_length:
            print(f"\n\U0001F973 BINGO! You've cracked the secret code in {max_attempts-steps+1} steps!")
            break
        
        # otherwise: give feedback
        steps -= 1
        print("\nFeedback:", end=" ")
        if feedback["black"] == 0 and feedback["white"] == 0:
            print("no matches")
        else:
            for peg in feedback.keys():
                for i in range(feedback[peg]):
                    print(fb_colors[peg], end="")
        print(f"\n{steps} attempts left.")

        # all attempts used? Too bad :(
        if steps == 0:
            print("\nALAS! \U0001F97A The secret code was:", end=" ")
            reveal_colors(sec_code, all_colors)
```

```python
# The one that starts them all

code_length = 4
num_attempts = 12
play_mastermind(code_length, num_attempts)
```
