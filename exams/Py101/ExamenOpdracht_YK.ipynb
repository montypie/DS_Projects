Examen opdracht: SHOPPING CART 


DESCRIPTION:
Create a shopping cart system with product inventory and receipt
generation. The system should handle multiple products and calculate totals with
taxes.

REQUIREMENTS:
1. Store product catalog in a dictionary:
   - Product ID as key
   - Values: name, price, category, stock_quantity
2. Store shopping cart as a dictionary:
   - Product ID as key
   - Value: quantity
3. Implement the following functions, add parameters as needed:
   - add_product_to_catalog(): Add product to inventory
            return confirmation if added, error message if ID not exists
   - add_to_cart(): Add a certain quantity of an item to shopping cart
            return confirmation if added, error message if insufficient stock or invalid product
   - remove_from_cart(): Remove item from cart
            return confirmation if removed, error message if not found
   - update_quantity(): Change quantity in cart
            return confirmation if updated, error message if insufficient stock or invalid product
   - view_cart(): Display cart contents and subtotal per product
            example output:
Your Cart:
Product A (x2) ........... €19.98 
Product B (x1) ............. €9.99

   - calculate_total(): Calculate cart with tax (21%)
            example output:
Subtotal: €29.97
TAX (21%): €6.29
TOTAL: €36.26


Example flow:

    add_product_to_catalog(catalog, 1, "Product A", 9.99, "Category 1", 100)
    -> "Product 'Product A' added to catalog!"

    add_product_to_catalog(catalog, 2, "Product B", 19.99, "Category 2", 50)
    -> "Product 'Product B' added to catalog!"

    add_to_cart(catalog, cart, 1, 2)
    -> "Added 2x 'Product A' to cart"

    add_to_cart(catalog, cart, 2, 1)
    -> "Added 1x 'Product B' to cart"

    remove_from_cart(cart, 1)
    -> "Product removed from cart"

    update_quantity(catalog, cart, 2, 3)
    -> "Quantity updated"

    view_cart(catalog, cart)
    ->  Your Cart:
        Product B (x3) ............. €59.97
        
    calculate_total(catalog, cart)
    ->  Subtotal: €59.97
        TAX (21%): €12.59
        TOTAL: €72.56

```python
def add_product_to_catalog(catalog, prod_id, name, price, category, stock_quantity):
    """ Add product to inventory """

    if prod_id in catalog.keys():
        error_msg = (f"ERROR: product with id {prod_id} already exists in the catalog.")
        return error_msg
    
    catalog[prod_id] = {"name": name,
                        "price": price,
                        "category": category,
                        "in_stock": stock_quantity
                        }
    ok_msg = (f"Product '{name}' is added to catalog!")

    return ok_msg

# two additional functions to check availability of a product in catalog and cart
def check_stock(catalog, prod_id, qnty):
    """ Check product availability """

    err = ""
    if prod_id not in catalog.keys():
        err = f"ERROR: no product with id {prod_id} in the catalog."
    elif qnty > catalog[prod_id]["in_stock"]:
        err = f"ERROR: insufficient quantity of such items in stock."
    return err

def check_cart(prod_id, cart):
    """ Check if product is in cart """

    if prod_id not in cart.keys():
        err = f"ERROR: no product with id {prod_id} in the cart."
        return err

def add_to_cart(catalog, cart, prod_id, qnty):
    """ Add a certain quantity of an item to shopping cart """

    error_msg = check_stock(catalog, prod_id, qnty)
    if error_msg:
        return error_msg
    
    cart.update({prod_id: qnty})
    ok_msg = (f"Added {qnty}x '{catalog[prod_id]["name"]}' to cart!")
    return ok_msg

def remove_from_cart(cart, prod_id):
    """ Remove item from cart """

    error_msg = check_cart(prod_id, cart)
    if error_msg:
        return error_msg

    del cart[prod_id]
    ok_msg = (f"Product with id {prod_id} removed from cart!")
    return ok_msg
    
def update_quantity(catalog, cart, prod_id, qnty):
    """ Change quantity in cart """

    error_msg = check_cart(prod_id, cart)
    if not error_msg:
        error_msg = check_stock(catalog, prod_id, qnty)
    if error_msg:
        return error_msg

    cart[prod_id] = qnty
    ok_msg = (f"Quantity for product with id {prod_id} adjusted in cart!")
    return ok_msg

def view_cart(catalog, cart):
    """ Display cart contents and subtotal per product """

    print("Your cart:")

    for prod, qnty in cart.items():
        item_str = f"{catalog[prod]["name"]} (x{qnty})"
        subtotal = catalog[prod]["price"] * qnty
        print(f"{item_str}) .................. €{subtotal}")

def calculate_total(catalog, cart):
    """ Calculate cart with tax (21%) """

    subtotal = 0
    for prod, qnty in cart.items():
        subtotal += catalog[prod]["price"] * qnty
    tax = subtotal / 100 * 21
    total = subtotal + tax

    print(f"Subtotal: €{subtotal}")
    print(f"TAX (21%): €{tax:.2f}")
    print(f"TOTAL: €{total:.2f}")
```
