### EDA and Plotting ###

```python
from sklearn import datasets
import pandas as pd

iris = datasets.load_iris()
# Create DataFrame for features
X = pd.DataFrame(iris.data, columns=iris.feature_names)
# Create Series for target
y = pd.Series(iris.target, name="species")
# Combine into one DataFrame (optional)
iris_df = pd.concat([X, y], axis=1)

print(iris_df.info())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Scatter plot using seaborn
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=iris_df,
    x='sepal length (cm)', 
    y='petal length (cm)', 
    size='petal width (cm)',  # Size of points by petal width
    hue='species',
    palette='pastel',
    edgecolor='black'
)
plt.title("Iris Species Scatter Plot")  # Set title
plt.xlabel("Sepal Length (cm)")  # Label x-axis
plt.ylabel("Petal Length (cm)")  # Label y-axis
plt.legend(title="Species")  # Improve legend title
plt.grid(True, linestyle='--', alpha=0.5)  # Add a grid for readability
plt.show()  # Display the plot
```

### Decision Tree ###

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

# Load the dataset
iris = load_iris()
X, y = iris.data, iris.target

# Train a Decision Tree Classifier
clf = DecisionTreeClassifier(max_depth=3, random_state=42)
clf.fit(X, y)

# Plot the decision tree
plt.figure(figsize=(10, 8))
plot_tree(clf, filled=True)
plt.show()
```

#### Linear Regression and SVM ###

```python
from sklearn.datasets import load_iris
import pandas as pd
import seaborn as sns

iris = load_iris(as_frame=True)
iris_df = iris.frame 

iris_df["species"] = iris_df["target"].map(dict(enumerate(iris.target_names)))
# Filtering 2 species for illustrating classification:
df_subset = iris_df[iris_df['species'].isin(['setosa', 'versicolor'])]

sns.scatterplot(data=df_subset, y="petal width (cm)", x="sepal width (cm)", hue="species")
```

```python
from sklearn.linear_model import LogisticRegression

X1=df_subset[['sepal width (cm)', 'petal width (cm)']]
y1=df_subset['species']

# Train logistic regression model
lrm1 = LogisticRegression()
lrm1.fit(X1,y1)
```

```python
from sklearn.svm import SVC

# Train SVM model
svm1 = SVC(kernel='linear')
svm1.fit(X1,y1)
```

```python
# Compare accuracies
print(f"Lin regression score: {lrm1.score(X1,y1)}")
print(f"SVM model score: {svm1.score(X1,y1)}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Plotting the iris data subset with decision boundaries
sns.scatterplot(data=df_subset, y="petal width (cm)", x="sepal width (cm)", hue="species")

# Logistic regression Decision Boundary line
w0_logreg = lrm1.intercept_[0]
w1_logreg, w2_logreg = lrm1.coef_[0]
x_vals_logreg = np.linspace(    X1["sepal width (cm)"].min() - 0.5,
                                X1["sepal width (cm)"].max() + 0.5,
                                100)

y_vals_logreg = -(w0_logreg + w1_logreg * x_vals_logreg) / w2_logreg
plt.plot(x_vals_logreg, y_vals_logreg, color='red', linestyle='-', linewidth=2,
         label='Decision Boundary LR')

# SVM Decision Boundary line
w0_svm = svm1.intercept_[0]
w1_svm , w2_svm  = svm1.coef_[0]
x_vals_svm  = np.linspace(  X1["sepal width (cm)"].min() - 0.5,
                            X1["sepal width (cm)"].max() + 0.5,
                            100)

y_vals_svm  = -(w0_svm  + w1_svm  * x_vals_svm ) / w2_svm
plt.plot(x_vals_svm , y_vals_svm , color='blue', linestyle='--', linewidth=2,label='Decision Boundary SVM')
```

```python
# Add new data point
new_row = pd.DataFrame({
    'sepal length (cm)': [5.1],
    'sepal width (cm)': [3.0],
    'petal length (cm)': [1.5],
    'petal width (cm)': [1.0],
    'species': ['setosa']
})
df_subset = pd.concat([df_subset, new_row], ignore_index=True)
sns.scatterplot(data=df_subset, y="petal width (cm)", x="sepal width (cm)", hue="species")
```

```python
# Re-train both models with a new data point

X2=df_subset[['sepal width (cm)', 'petal width (cm)']].values
y2=df_subset['species']

lrm2 = LogisticRegression(C=40.0, solver="liblinear") # additional parameters for better fitting
lrm2.fit(X2,y2)
svm2 = SVC(kernel='linear')
svm2.fit(X2,y2)

# Compare new accuracies
print(f"Lin regression plus score: {lrm2.score(X2,y2):.2f}")
print(f"SVM model plus score: {svm2.score(X2,y2):.2f}")
```

```python
# Re-plotting the updated iris data subset with decision boundaries
sns.scatterplot(data=df_subset, y="petal width (cm)", x="sepal width (cm)", hue="species")

# Logistic regression Decision Boundary line
w0_logreg = lrm2.intercept_[0]
w1_logreg, w2_logreg = lrm2.coef_[0]
x_vals_logreg = np.linspace(    X1["sepal width (cm)"].min() - 0.5,
                                X1["sepal width (cm)"].max() + 0.5,
                                100)

y_vals_logreg = -(w0_logreg + w1_logreg * x_vals_logreg) / w2_logreg
plt.plot(x_vals_logreg, y_vals_logreg, color='red', linestyle='-', linewidth=2,
         label='Decision Boundary Logistic')

# SVM Decision Boundary line
w0_svm = svm2.intercept_[0]
w1_svm , w2_svm  = svm2.coef_[0]
x_vals_svm  = np.linspace(  X1["sepal width (cm)"].min() - 0.5,
                            X1["sepal width (cm)"].max() + 0.5,
                            100)

y_vals_svm  = -(w0_svm  + w1_svm  * x_vals_svm ) / w2_svm
plt.plot(x_vals_svm , y_vals_svm , color='blue', linestyle='--', linewidth=2,label='Decision Boundary SVM')
```
